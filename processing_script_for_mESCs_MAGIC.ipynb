```python
df = pd.read_pickle("scTech_datasets/adata_wgx.p")
df.shape
```

```python
annotation = pd.read_csv("./scTech_datasets/Mouse_sctechnologies.annotation.tsv", sep='\t')
annotation.head()
```

```python
df_magic = []
for Tech in np.unique(annotation.Cluster.values):
    df_adata = df.iloc[:, annotation.Cluster.values==Tech].T
    
    if Tech == "BulkSeq":
        pass #df_magic.append(df_adata)
    else:
        magic_op = magic.MAGIC()
        df_adata_magic = magic_op.fit_transform(df_adata)

        print(df_adata_magic.shape)
        df_magic.append(df_adata_magic)
        
df_magic = pd.concat(df_magic, axis=0).T
df_magic = df_magic[annotation.iloc[2:,].Cell] # sort df_magic cell_id
```

```python
annotation_magic = annotation.iloc[2:,].copy() # omit bulk
annotation_magic.Cell = annotation.Cell.apply(lambda x:"MAGIC-"+x)
annotation_magic.Species = annotation.Species.apply(lambda x:"MAGIC-"+x)
annotation_magic.Celltype = annotation.Celltype.apply(lambda x:"MAGIC-"+x)
annotation_magic.Cluster = annotation.Cluster.apply(lambda x:"MAGIC-"+x)
annotation_magic.head()
```

```python
np.unique(annotation_magic.Cluster, return_counts=True)
```

```python
annotation_magic.to_csv("./Mouse_sctechnologies_MAGIC.annotation_wgx.tsv", sep='\t', index=None)
```
